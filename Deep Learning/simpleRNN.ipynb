```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Load IMDB dataset
max_words = 10000  # Top 10,000 most frequent words
max_len = 500      # Max length of each review

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_words)

# Pad sequences to the same length
x_train = pad_sequences(x_train, maxlen=max_len)
x_test = pad_sequences(x_test, maxlen=max_len)

# Build the RNN model
model = Sequential([
    Embedding(input_dim=max_words, output_dim=32, input_length=max_len),
    SimpleRNN(units=32),
    Dense(1, activation='sigmoid')  # Output: probability between 0 and 1
])

# Compile the model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train, epochs=3, batch_size=64, validation_split=0.2)

# Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test)
print("\nTest Accuracy:", test_acc)

# Make predictions on the first 5 test reviews
predictions = model.predict(x_test[:5])

# Print predicted vs actual sentiment
for i in range(5):
    prob = predictions[i][0]
    predicted_label = "Positive" if prob > 0.5 else "Negative"
    actual_label = "Positive" if y_test[i] == 1 else "Negative"

    print(f"\nReview {i+1}:")
    print(f"Predicted Probability: {prob:.4f}")
    print(f"Predicted Sentiment: {predicted_label}")
    print(f"Actual Label: {actual_label}")
```
