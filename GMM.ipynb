```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.mixture import GaussianMixture

# Load the iris dataset
iris = datasets.load_iris()
X = iris.data  # Features
y = iris.target  # Target variable (not used in clustering)

# Define the number of iterations
n_iter = 10

# Fit a Gaussian Mixture Model with three components
gmm = GaussianMixture(n_components=3)


# Initialize figure for plotting
plt.figure(figsize=(8, 6))

for i in range(n_iter):
    # Fit the model
    gmm.fit(X)

    # Predicted labels from GMM
    labels = gmm.predict(X)

    # Plot each of the three clusters with a different color
    for j in range(3):
        plt.scatter(X[labels == j, 0], X[labels == j, 1], label=f'Cluster {j+1}')

    # Plotting the centroids of the clusters
    plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], color='black', marker='*', s=200, label='Centroids')

    plt.title(f'Iteration {i+1}')
    plt.xlabel('Sepal Length')
    plt.ylabel('Sepal Width')
    plt.legend()

    # Pause for a moment to allow time for visualization
    plt.pause(0.5)
    plt.show()
# plt.tight_layout()

```
