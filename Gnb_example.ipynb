```python
# load the iris dataset
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
# store the feature matrix (X) and response vector (y)
X = iris.data
y = iris.target
```

```python
# splitting X and y into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=1)
```

```python

# training the model on training set
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
gnb.fit(X_train, y_train)
```

```python
# making predictions on the testing set
y_pred = gnb.predict(X_test)
```

```python
# comparing actual response values (y_test) with predicted response values (y_pred)
from sklearn import metrics
print("Gaussian Naive Bayes model accuracy(in %):", metrics.accuracy_score(y_test, y_pred)*100)
```

```python
import pandas as pd
import numpy as np
adultdata = pd.read_csv("adult.csv", header=None, sep=',\s')
```

### Rename column names
We can see that the dataset does not have proper column names. The columns are merely labelled as 0,1,2.... and so on. We should give proper names to the columns. I will do it as follows:-

```python
col_names = ['age', 'workclass', 'fnlwgt', 'education', 'education_num', 'marital_status', 'occupation', 'relationship',
             'race', 'sex', 'capital_gain', 'capital_loss', 'hours_per_week', 'native_country', 'income']

adultdata.columns = col_names

adultdata.columns
```

```python
adultdata.head()
```

We can see that the column names are renamed. Now, the columns have meaningful names.

### View summary of dataset

```python
adultdata.info()
```

We can see that there are no missing values in the dataset.

### Types of variables
Here we can segregate the dataset into categorical and numerical variables. There are a mixture of categorical and numerical variables in the dataset. Categorical variables have data type object. Numerical variables have data type int64.

First of all, let's explore categorical variables.

```python
# find categorical variables

categorical = [var for var in adultdata.columns if adultdata[var].dtype=='O']

print('There are {} categorical variables\n'.format(len(categorical)))

print('The categorical variables are :\n\n', categorical)
```

```python
# view the categorical variables

adultdata[categorical].head()
```

### Summary of categorical variables

* There are 9 categorical variables.
* The categorical variables are given by workclass, education, marital_status, occupation, relationship, race, sex, native_country and income.
* income is the target variable.

```python
# check missing values in categorical variables

adultdata[categorical].isnull().sum()
```

```python
# view frequency counts of values in categorical variables

for var in categorical:

    print(adultdata[var].value_counts())
```

Now, we can see that there are several variables like workclass, occupation and native_country which contain missing values. Generally, the missing values are coded as NaN and python will detect them with the usual command of df.isnull().sum().

But, in this case the missing values are coded as ?. Python fail to detect these as missing values because it do not consider ? as missing values. So, I have to replace ? with NaN so that Python can detect these missing values.

let's explore these variables and replace ? with NaN.

```python
# check labels in workclass variable

adultdata.workclass.unique()
```

```python
# check frequency distribution of values in workclass variable

adultdata.workclass.value_counts()
```

```python
# replace '?' values in workclass variable with `NaN`


adultdata['workclass'].replace('?', np.nan, inplace=True)
```

```python
# again check the frequency distribution of values in workclass variable

adultdata.workclass.value_counts()
```

```python
# check labels in occupation variable

adultdata.occupation.unique()
```

```python
# check frequency distribution of values in occupation variable

adultdata.occupation.value_counts()
```

```python
# replace '?' values in occupation variable with `NaN`

adultdata['occupation'].replace('?', np.nan, inplace=True)
```

```python
adultdata.occupation.value_counts()
```

```python
# check labels in native_country variable

adultdata.native_country.unique()
```

```python
# check frequency distribution of values in native_country variable

adultdata.native_country.value_counts()
```

```python
# replace '?' values in native_country variable with `NaN`

adultdata['native_country'].replace('?', np.nan, inplace=True)
```

```python
adultdata.native_country.value_counts()
```

```python
adultdata[categorical].isnull().sum()
```

Now, we can see that workclass, occupation and native_country variable contains missing values.

```python
adultdata.income.value_counts()
```

```python
adultdata.dropna(inplace=True)
```

```python
adultdata[categorical].isnull().sum()
```

```python
# from sklearn.preprocessing import LabelEncoder
# le = LabelEncoder()
# for var in categorical:
#     adultdata[var] = le.fit_transform(adultdata[var])


from sklearn.preprocessing import LabelEncoder

categorical = adultdata.select_dtypes(include='object').columns

for var in categorical:
    adultdata[var] = LabelEncoder().fit_transform(adultdata[var])
```

```python
adultdata.head()
```

```python
adultdata
```

```python
X = adultdata.drop(['income'], axis=1)

y = adultdata['income']
```

```python
# splitting X and y into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
gnb.fit(X_train, y_train)
```

```python
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```
