```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv("income.data.csv")
df
```

Show the Relationship between the income and the happiness:

```python
plt.scatter(df.income, df.happiness)
plt.title("The Correlation between Income and Happiness")
plt.ylabel("Happiness")
plt.xlabel("Income")
plt.show()
```

We found that based on the previous plot, whenever the Income increases the Happiness Increases.

Use Simple Linear Regression to Predict the Happiness based on the Income:

Now we will create a model to predict the Happiness rate. The function of the model will be in this form :

Predicted_Happiness = W * Income + B

or we can write in easier way like this:

f(x) = w * x + b

where:

f(x) ==> Predicted value of Happiness
x ==> Value of Income
w, b ==> model parameters

```python
X = pd.DataFrame(df["income"])
Y = pd.DataFrame(df["happiness"])
```

```python
print(X.shape,Y.shape)
```

```python
X_train , X_test , Y_train , Y_test = train_test_split(X,Y, test_size = 0.30 , random_state = 2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
print(Y.shape,Y_train.shape,Y_test.shape)
```

```python
lin = LinearRegression()
lin.fit(X_train,Y_train)
```

```python
y_pred = lin.predict(X_test)
y_pred
```

```python
print(r2_score(Y_test,y_pred)*100)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

dataset = pd.read_csv('Salary_Data.csv')
dataset.head()
```

```python
X = dataset.iloc[:,:-1].values  #independent variable array
y = dataset.iloc[:,1].values  #dependent variable vector
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train,y_train)
```

```python
y_pred = regressor.predict(X_test)
y_pred
```

```python
#plot for the TRAIN

plt.scatter(X_train, y_train, color='red') # plotting the observation line

plt.plot(X_train, regressor.predict(X_train), color='blue') # plotting the regression line

plt.title("Salary vs Experience (Training set)") # stating the title of the graph

plt.xlabel("Years of experience") # adding the name of x-axis
plt.ylabel("Salaries") # adding the name of y-axis
plt.show() # specifies end of graph
```

```python
#plot for the TEST

plt.scatter(X_test, y_test, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue') # plotting the regression line

plt.title("Salary vs Experience (Testing set)")

plt.xlabel("Years of experience")
plt.ylabel("Salaries")
plt.show()
```
